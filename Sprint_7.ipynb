### Segundo desgranaje:

✔ `Validar datos`

    *  Recibir los seis datos introducidos.
    *  Validar que el nombre y los apellidos no estén vacíos.
    *  Validar que el teléfono tenga nueve caracteres y que todos sean dígitos.
    *  Validar que los años de experiencia sean un número entero igual o superior a cero.
    *  Validar que catalán, Excel y titulación sean números enteros comprendidos entre 0 y 3.
    *  Si todos los datos cumplen las condiciones, devolver True.
    *  Si alguno de los datos es incorrecto, mostrar un mensaje de error y devolver False.

3. Calcular su puntuación individual y mostrárselo.
4. Clasificar al candidato por prioridad según su puntuación.
5. Mostrar al candidato su puntuación individual, incluyendo su nombre, puntuación y clasificación.

```python
def validar_datos (
    nombre_apellidos:str,
    telefono_contacto:str,
    años_experiencia: str,
    nivel_catalan:str,
    dominio_excel:str,
    titulacion_academica:str
) -> bool:
    # Validar el nombre y apellidos

    if nombre_apellidos.strip() == "":
      print('Error: el nombre y los apellidos no pueden estar vacíos.')
      return False

    # Validar el teléfono

    if not telefono_contacto.isdigit() or len(telefono_contacto) != 9:
      print('Error: el teléfono debe tener nueve dígitos.')
      return False

    # Intentar convertir los siguientes datos a enteros

    try:
      años_experiencia = int(años_experiencia)
      nivel_catalan = int(nivel_catalan)
      dominio_excel = int(dominio_excel)
      titulacion_academica = int(titulacion_academica)
    except ValueError:
      print('Error: debe introducir números enteros')
      return False

    # Validar los años de experiencia que sea número entero igual o superior a cero.

    if años_experiencia < 0:
      print('Error: los años de experiencia deben ser un número entero igual o superior a cero.')
      return False

    # Validar que el nivel de catalán, el dominio de excel y titulación académica sean números enteros comprendidos entre 0 y 3

    if nivel_catalan < 0 or nivel_catalan > 3:
      print('Error: el nivel de catalán debe estar entre 0 y 3.')
      return False
    if dominio_excel < 0 or dominio_excel > 3:
      print('Error: el nivel de Excel debe estar entre 0 y 3.')
      return False
    if titulacion_academica < 0 or titulacion_academica > 3:
      print('Error: el nivel de titulación debe estar entre 0 y 3.')
      return False

    return True
```

✔ `Calcular Puntuación`

```python
def calcular_puntuacion(
    años_experiencia: str,
    nivel_catalan: str,
    dominio_excel: str,
    titulacion_academica: str
) -> float:
  años_experiencia = int(años_experiencia)
  nivel_catalan = int(nivel_catalan)
  dominio_excel = int(dominio_excel)
  titulacion_academica = int(titulacion_academica)

  puntuacion = años_experiencia * 1.5 + nivel_catalan + dominio_excel + titulacion_academica

  return puntuacion
```

✔ `Clasificar candidato`

```python
def clasificar_candidato(puntuacion: float) -> str:
  if puntuacion >= 12:
    return 'Prioridad Alta'
  elif puntuacion >= 8:
    return 'Candidato adecuado'
  else:
    return 'No prioritario'
```

### Sexto desgranaje

✔ `Mostrar informe`

    a. Recibir la lista de candidatos válidos.

    b. Ordenar los candidatos primero según su clasificación:

        Prioridad Alta.
        Candidato adecuado.
        No prioritario.

    c. Si varios candidatos tienen la misma clasificación, ordenarlos por puntuación de mayor a menor.

    d. Recorrer la lista de candidatos ya ordenada.

    e. Mostrar el nombre y apellidos, el teléfono, la puntuación y la clasificación de cada candidato.


7. Crear la función de orquestación main_seleccion().

```python
def mostrar_informe(lista_candidatos: list[dict]
) -> None:

  prioridades = {
      'Prioridad Alta': 3,
      'Candidato adecuado': 2,
      'No prioritario': 1
  }

  lista_candidatos.sort(
      key=lambda candidato: (
          prioridades[candidato['clasificacion']],
          candidato['puntuacion']
          ),
          reverse=True
 )

  print('\n---Informe final---')

  for candidato in lista_candidatos:
    print(f"Nombre: {candidato['nombre_apellidos']}")
    print(f"Teléfono: {candidato['telefono']}")
    print(f"Puntuación: {candidato['puntuacion']}")
    print(f"Clasificación: {candidato['clasificacion']}")
    print()
```

### Séptimo desgranaje:

✔ `Crear la función de orquestación` - main_seleccion().

    a. Crear una lista vacía para guardar a los candidatos.

    b. Repetir el proceso mientras la lista contenga menos de diez candidatos válidos.

    c. Llamar a obtener_datos() para solicitar los seis datos de una persona.

    d. Llamar a validar_datos() para comprobar que los datos introducidos sean correctos.

    e. Si validar_datos() devuelve False, no guardar al candidato y volver al inicio del bucle para solicitar nuevos datos.

    f. Si los datos son válidos, llamar a calcular_puntuacion().

    g. Llamar a clasificar_candidato() utilizando la puntuación obtenida.

    h. Crear un diccionario con el nombre y apellidos, el teléfono, la puntuación y la clasificación.

    i. Añadir el diccionario a la lista de candidatos.

    j. Llamar a mostrar_individual() para mostrar el resultado de la persona registrada.

    k. Cuando se hayan registrado diez candidatos válidos, finalizar el bucle.

    l. Llamar a mostrar_informe() y entregarle la lista de candidatos.

```python
def main_seleccion() -> None:

    # Crear la lista donde se guardarán los candidatos válidos
    lista_candidatos = []

    # Repetir el proceso hasta registrar diez candidatos válidos
    while len(lista_candidatos) < 10:

        # Solicitar los datos de una persona
        (
            nombre_apellidos,
            telefono_contacto,
            años_experiencia,
            nivel_catalan,
            dominio_excel,
            titulacion_academica
        ) = obtener_datos()

        # Comprobar si los datos introducidos son válidos
        resultado = validar_datos(
            nombre_apellidos,
            telefono_contacto,
            años_experiencia,
            nivel_catalan,
            dominio_excel,
            titulacion_academica
        )

        # Procesar únicamente a los candidatos con datos válidos
        if resultado:

            # Calcular la puntuación del candidato
            puntuacion = calcular_puntuacion(
                años_experiencia,
                nivel_catalan,
                dominio_excel,
                titulacion_academica
            )

            # Obtener la clasificación correspondiente
            clasificacion = clasificar_candidato(puntuacion)

            # Reunir los resultados del candidato en un diccionario
            candidato = {
                "nombre_apellidos": nombre_apellidos,
                "telefono": telefono_contacto,
                "puntuacion": puntuacion,
                "clasificacion": clasificacion
            }

            # Guardar al candidato válido en la lista
            lista_candidatos.append(candidato)

            # Mostrar el resultado individual
            mostrar_individual(
                nombre_apellidos,
                puntuacion,
                clasificacion
            )

    # Mostrar el informe cuando se hayan registrado diez candidatos
    mostrar_informe(lista_candidatos)
```

```python
#Ponemos como parámetro "codigo_comportamiento" porque la función necesita saber qué comportamiento debe consultar para cada cliente.
def obtener_segmento_y_factor(codigo_comportamiento: int) -> dict[str, str | float]:
  comportamientos = {
      0: {
            'segmento': 'Navega pero no añade productos',
            'mensaje': 'Descubre nuestras novedades y déjate inspirar',
            'factor': 0.10
          },
      1: {
            'segmento': 'Agrega productos pero no llega al checkout',
            'mensaje': 'Tienes productos esperándote, complétalos ahora',
            'factor': 0.25
          },
      2: {
            'segmento': 'Llega al checkout pero no paga',
            'mensaje': 'Estás a un paso de tu compra, ¿te podemos ayudar en algo?',
              'factor': 0.50
          },
      3: {
            'segmento': 'Cesta de compra abandonada hace menos de 24 h',
            'mensaje': 'Tu cesta de compras te está esperando',
            'factor': 0.75
          },
      4: {
            'segmento': 'Cesta de compra abandonada hace más de 24 h',
            'mensaje': 'Te echamos de menos, recupera tu cesta y termina tu compra',
            'factor': 0.90
          }
  }
  return comportamientos[codigo_comportamiento]
```

### Segundo desgranaje

✔ `Obtener factor edad`

  *   Consultar el factor correspondiente al grupo de edad del cliente.
  *   Crear un diccionario para poder agrupar los grupos de edades (que serían las claves) y su factor.
  *   Devolver el factor edad según el grupo de edad.

```python
def obtener_factor_edad(grupo_edad: str) -> float:
  edades = {
      '18-25': 1.20,
      '26-40': 1.00,
      '41-60': 0.85,
      '60+': 0.70
  }
  return edades[grupo_edad]
```

### Tercer desgranaje

✔ `Calcular riesgo`

  *   Recibir el factor de comportamiento y el factor de edad.
  *   Multiplicar el factor de comportamiento por el factor de edad.
  *   Devolver el riesgo calculado

```python
def calcular_riesgo(
    factor_comportamiento: float,
    factor_edad: float
) -> float:

    return factor_comportamiento * factor_edad
```

### Cuarto desgranaje

✔ `Procesar cliente`

  *   Recibir la lista de clientes.
  *   Crear una lista vacía para guardar los resultados.
  *   Recorrer los clientes.
  *   Obtener el segmento, mensaje y factor de comportamiento.
  *   Calcular el riesgo con ambos factores.
  *   Guardar los datos originales y los resultados en un diccionario.
  *   Añadir ese diccionario a la lista de resultados.
  *   Devolver la lista al terminar.

```python
def procesar_clientes(
    clientes: list[dict]
) -> list[dict]:

  # Crear la lista donde se guardarán los clientes procesados
    clientes_procesados = []

    # Procesar los clientes uno por uno
    for cliente in clientes:

        # Buscar los datos correspondientes al código de comportamiento del cliente
        datos_comportamiento = obtener_segmento_y_factor(
            cliente["codigo_comportamiento"]
        )

        # Extraer el factor que se utilizará para calcular el riesgo
        factor_comportamiento = datos_comportamiento["factor"]

        # Obtener el factor correspondiente al grupo de edad
        factor_edad = obtener_factor_edad(
            cliente["grupo_edad"]
        )

        # Calcular el riesgo final del cliente
        riesgo = calcular_riesgo(
            factor_comportamiento,
            factor_edad
        )

        # Crear un nuevo diccionario con toda la información del cliente
        cliente_procesado = {
            "id_cliente": cliente["id_cliente"],
            "nombre": cliente["nombre"],
            "edad": cliente["edad"],
            "grupo_edad": cliente["grupo_edad"],
            "codigo_comportamiento": cliente["codigo_comportamiento"],
            "segmento": datos_comportamiento["segmento"],
            "mensaje": datos_comportamiento["mensaje"],
            "factor_comportamiento": factor_comportamiento,
            "factor_edad": factor_edad,
            "riesgo": riesgo
        }

        # Añadir el cliente procesado a la nueva lista
        clientes_procesados.append(cliente_procesado)

    return clientes_procesados
```

### Primer desgranaje

*Separar y normalizar la lista de síntomas — objetivo 1*

Entrada: la función recibe mediante "cadena_sintomas" una cadena de texto que contiene los síntomas separados por comas.

Salida: devuelve una lista con los síntomas separados y normalizados.

1. Separar "cadena_sintomas" utilizando la coma como separador, con split().

2. Guardar la lista resultante en "sintomas_sin_normalizar".

3. Crear una lista vacía llamada "sintomas_limpios".

4. Recorrer mediante un bucle cada elemento de "sintomas_sin_normalizar". En cada vuelta, la variable "sintoma" representará un síntoma.

5. Para cada síntoma:

    a. Eliminar los espacios sobrantes del principio y del final mediante strip().

    b. Convertir el síntoma a minúsculas mediante lower().

    c. Reemplazar los espacios interiores por guiones bajos mediante replace(" ", "_").

    d. Guardar el resultado en "sintoma_normalizado".

    e. Añadir "sintoma_normalizado" a la lista sintomas_limpios.

6. Cuando se hayan procesado todos los síntomas, devolver la lista sintomas_limpios.


```python
# --- 1. FUNCIÓN PARA SEPARAR Y NORMALIZAR SÍNTOMAS ---
def normalizar_sintomas(cadena_sintomas: str) -> list:

    sintomas_sin_normalizar = cadena_sintomas.split(",")

    sintomas_limpios = []

    for sintoma in sintomas_sin_normalizar:

        sintoma_normalizado = (
            sintoma.strip()
            .lower()
            .replace(" ", "_")
        )

        sintomas_limpios.append(sintoma_normalizado)

    return sintomas_limpios
```

### Segundo desgranaje

* ***Convertir cada string en una estructura limpia - objetivo 2*** *

Entrada: la función recibe la lista pacientes, donde cada paciente está representado por una cadena de texto.

Salida: devuelve una lista de diccionarios con los pacientes estructurados.

1. Crear una lista vacía llamada "pacientes_estructurados", donde se guardarán los diccionarios de los pacientes.

2. Recorrer mediante un bucle cada elemento de la lista "pacientes". En cada vuelta, la variable "paciente" representará la cadena de texto de un paciente.

3. Para cada paciente:

    a. Separar la cadena de texto utilizando | como separador. El método split() crea una lista de tres elementos y la guarda en "datos_paciente".

    b. Obtener el nombre, situado en la primera posición [0] de "datos_paciente". Eliminar sus espacios sobrantes mediante strip() y guardar el resultado en nombre.

    c. Obtener la edad, situada en la segunda posición [1]. Convertirla a número entero mediante int() y guardar el resultado en edad.

    d. Obtener el bloque de síntomas, situado en la tercera posición [2], y guardarlo en cadena_sintomas.

    e. Enviar "cadena_sintomas" a la función definida anteriormente: "normalizar_sintomas()".

    f. Guardar la lista devuelta por normalizar_sintomas() en "lista_sintomas_limpios".

    g. Crear un diccionario llamado "paciente_diccionario" con el nombre, la edad y la lista de síntomas limpios.

    h. Añadir "paciente_diccionario" a la lista "pacientes_estructurados".

4. Cuando se hayan procesado todos los pacientes, devolver la lista "pacientes_estructurados".

```python
# --- 2. FUNCIÓN PARA CONVERTIR STRINGS EN DICCIONARIOS ---
def estructurar_pacientes(pacientes: list) -> list:

    pacientes_estructurados = []

    for paciente in pacientes:
        datos_paciente = paciente.split("|")

        nombre = datos_paciente[0].strip()
        edad = int(datos_paciente[1])
        cadena_sintomas = datos_paciente[2]

        # Llama a la función normalizar_sintomas()
        lista_sintomas_limpios = normalizar_sintomas(
            cadena_sintomas
        )

        paciente_diccionario = {
            "nombre": nombre,
            "edad": edad,
            "sintomas": lista_sintomas_limpios
        }

        pacientes_estructurados.append(
            paciente_diccionario
        )

    return pacientes_estructurados
```

### Cuarto desgranaje

* ***Calcular el riesgo total*** *

Entrada: la lista de síntomas de un paciente y el diccionario niveles_riesgo.

Salida: el riesgo total del paciente como número entero.

1. Crear un acumulador llamado riesgo_total que comience en 0.

2. Recorrer uno a uno los elementos de la lista "sintomas.

3. En un bucle, en cada vuelta, enviar el sintoma actual y el diccionario "niveles_riesgo" a la función "asignar_riesgo_sintoma()"

4. Guardar el número devuelto en la variable "riesgo_sintoma"

5. Sumar riesgo_sintoma al acumulador "riesgo_total".

6. Cuando se hayan recorrido todos los sintomas, devolver "riesgo_total".

```python
def calcular_riesgo_total(sintomas: list, niveles_riesgo: dict) -> int:

  riesgo_total = 0

  for sintoma in sintomas:

    riesgo_sintoma = asignar_riesgo_sintoma(sintoma, niveles_riesgo)
    riesgo_total += riesgo_sintoma

  return riesgo_total
```

    3. La categoria obtenida anteriormente mediante el parámetro "categoria", de tipo (str).

    4. El precio base de la póliza mediante el parámetro "precio_base", de tipo (float).

    5. El precio final de la póliza calculado anteriormente mediante el parámetro "precio_final_poliza", de tipo (float).

    6. La probabilidad de fraude mediante el parámetro "probabilidad_fraude", de tipo (float).

Salida:

    Devolverá un diccionario (dict) llamado "diccionario_final_cliente" con la información final de un cliente.

Proceso:

    1. Crear el diccionario "diccionario_final_cliente" con las siguientes claves y sus correspondientes variables:

    *"nombre": nombre
    *"edad": edad
    *"categoria": categoria
    *"precio_base": precio_base
    *"precio_final_poliza": precio_final_poliza
    *"probabilidad:fraude": probabilidad_fraude

    2. Devolver diccionario_final_cliente.


  





```python
def crear_diccionario_final_cliente(
    nombre:str,
    edad:int,
    categoria:str,
    precio_base:float,
    precio_final_poliza:float,
    probabilidad_fraude:float
    ) -> dict:

    diccionario_final_cliente = {
        "nombre": nombre,
        "edad": edad,
        "categoria": categoria,
        "precio_base": precio_base,
        "precio_final_poliza": precio_final_poliza,
        "probabilidad_fraude": probabilidad_fraude
    }
    return diccionario_final_cliente
```

### Sexto desgranaje

Procesar todos los clientes y guardar su infomación final.

Entrada: la función recibirá:

    La lista "clientes".

    El diccionario "niveles_seguro".

Salida:

    Devolverá una lista llamada "clientes_procesados", que contendrá el diccionario final de cada cliente.

Proceso:

    1. Crear una lista vacía llamada "clientes_procesados".

    2. Recorrer la lista "clientes" mediante un bucle. En cada vuelta, la variable "cliente" representará al diccionario de un cliente.

    3. Obtener del diccionario "cliente":

      * El nombre mediante la clave "nombre".
      * La edad mediante la clave "edad".
      * El código mediante la clave "codigo".

    4. Utilizar el código y las funciones anteriores para obtener:

      * La categoría.
      * El precio base.
      * La probabilidad de fraude.

    5. Calcular el precio final de la póliza usando el precio y la edad.

    6. Llamar a crear_diccionario_final_cliente() y pasarle todos los datos obtenidos.

    7. Añadir el diccionario resultante a la lista "clientes_procesados".

    8. Cuando termine el bucle, devolver "clientes_procesados".







```python
def procesar_clientes_seguros(
    clientes_seguros: list,
    niveles_seguro: dict
) -> list:

    # Crear una lista para guardar los clientes procesados
    clientes_procesados = []

    # Recorrer uno por uno los clientes de seguros
    for cliente in clientes_seguros:

        # Obtener los datos originales del cliente
        nombre = cliente["nombre"]
        edad = cliente["edad"]
        codigo = cliente["codigo"]

        # Consultar la información correspondiente a su código
        categoria = obtener_categoria(
            codigo,
            niveles_seguro
        )

        precio_base = obtener_precio_base(
            codigo,
            niveles_seguro
        )

        probabilidad_fraude = obtener_probabilidad_fraude(
            codigo,
            niveles_seguro
        )

        # Calcular el precio final de la póliza
        precio_final_poliza = calcular_precio_final_poliza(
            precio_base,
            edad
        )

        # Crear el diccionario con toda la información calculada
        diccionario_final_cliente = crear_diccionario_final_cliente(
            nombre,
            edad,
            categoria,
            precio_base,
            precio_final_poliza,
            probabilidad_fraude
        )

        # Añadir el cliente procesado a la lista final
        clientes_procesados.append(
            diccionario_final_cliente
        )

    return clientes_procesados
```

```python
#Comprobar la función procesar_clientes

clientes_procesados = procesar_clientes_seguros(
    clientes_seguros,
    niveles_seguro
)

for cliente in clientes_procesados:
  print(cliente)
```

### Desgranaje Once

Crear una función de orquestación que conecte todas las funciones anteriores.

    2. Llamar a ordenar_clientes_por_fraude() utilizando "clientes_procesados" y guardar el resultado en "clientes_ordenados_fraude".

### Primer desgranaje:

Validar la configuración recibida.

Entrada: Crear una función llamada "validar_configuracion" que reciba los 5 parámetros: longitud, mayusculas, minusculas, numeros y signos. Debe comprobar que:

    * La longitud sea un número entero mayor a cero.
    * Se haya seleccionado al menos un tipo de caracter.
    * La longitud sea igual o mayor que la cantidad de tipos seleccionados, para poder incluir como mínimo un caracter de cada tipo.

Salida: Si alguna condición no se cumple, se mostrará el mensaje de error correspondiente y la función devolverá False. Si todas las condiciones se cumplen, la función devolverá True y se podrá continuar con la generación de la contraseña.

```python
def validar_configuracion(
    longitud: int,
    mayusculas: bool,
    minusculas: bool,
    numeros: bool,
    signos: bool) -> bool:

  #Comprobar que la longitud sea un número entero mayor que cero

  if not isinstance(longitud, int) or longitud <= 0:
    print('Error: la longitud debe ser un número entero mayor que cero.')
    return False

  #Comprobar que se haya seleccionado al menos un tipo de caracter

  if not mayusculas and not minusculas and not numeros and not signos:
    print('Error: debe seleccionar al menos un tipo de caracter.')
    return False

  #Contar cuántos tipos de caracteres se han seleccionado
  cantidad_tipos = sum([mayusculas, minusculas, numeros, signos])

  #Comprobar que la longitud permita incluir un caracter de cada tipo
  if longitud < cantidad_tipos:
    print('Error: la longitud no permite incluir un caracter de cada tipo.')
    return False

  return True
```

### Segundo desgranaje:

Obtener los grupos seleccionados

Entrada:

    Crear una función que reciba los 4 parámetros booleanos de mayusculas, minusculas, numeros y signos.
    Cada parámetro indicará mediante True o False si ese grupo debe utilizarse para generar la contraseña.

Proceso:

    1. Definir una cadena de texto con todas las mayúsculas.

    2. Definir una cadena de texto con todas las minúsculas.

    3. Definir una cadena de texto con todas los números del 0 al 9.

    4. Definir una cadena de texto con los signos especiales permitidos.

    5. Crear una lista vacía llamada "grupos_seleccionados". En esta lista se guardarán únicamente los grupos cuyos parámetros tengan el valor True.

    6. Comprobar el parámetro mayusculas. Si contiene True, añadir la cadena de letras mayúsculas a la lista.

    7. Comprobar el parámetro minusculas. Si contiene True, añadir la cadena de letras minúsculas.

    8. Comprobar el parámetro numeros. Si contiene True, añadir la cadena de números.

    9. Comprobar el parámetro signos. Si contiene True, añadir la cadena de signos especiales.

    10. Devolver la lista con los grupos seleccionados.


```python
def obtener_grupos_caracteres(
    mayusculas: bool,
    minusculas: bool,
    numeros: bool,
    signos: bool
) -> list[str]:

    letras_mayusculas = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    letras_minusculas = "abcdefghijklmnopqrstuvwxyz"
    caracteres_numericos = "0123456789"
    caracteres_especiales = "!@#$%&*?+-_"

    grupos_seleccionados = []

    if mayusculas:
        grupos_seleccionados.append(letras_mayusculas)

    if minusculas:
        grupos_seleccionados.append(letras_minusculas)

    if numeros:
        grupos_seleccionados.append(caracteres_numericos)

    if signos:
        grupos_seleccionados.append(caracteres_especiales)

    return grupos_seleccionados
```

### Tercer desgranaje:

Seleccionar los caracteres obligatorios.

Entrada:

    Crear una función llamada seleccionar_caracteres_obligatorios() que reciba la lista de grupos creada por obtener_grupos_caracteres().

    Su objetivo será seleccionar aleatoriamente un carácter de cada grupo recibido. Así nos aseguramos de que la contraseña contenga, como mínimo, un carácter de cada tipo configurado.

Proceso:

    1. Importar NumPy as np antes de crear la función.

    2. La función recibirá "grupos_seleccionados", que será una lista de cadenas de texto.

    3. Crear una lista vacía llamada "caracteres_obligatorios". Aquí guardaremos un caracter de cada grupo.

    4. Recorrer los grupos seleccionados utilizando un bucle for. En cada vuelta, grupo tendrá un grupo diferente.

    5. Convertir cada grupo en una lista de caracteres. Esto permitirá que NumPy pueda escoger uno de los caracteres.

    6. Utilizar np.random.choice() para seleccionar aleatoriamente un carácter del grupo.

    7. Añadir el carácter elegido a la lista caracteres_obligatorios.

    8. Cuando el for haya recorrido todos los grupos, devolver la lista con los caracteres obligatorios.


```python
import numpy as np
```

```python
def seleccionar_caracteres_obligatorios(
    grupos_seleccionados: list[str]
) -> list[str]:

    caracteres_obligatorios = []

    for grupo in grupos_seleccionados:
        caracter_elegido = str(np.random.choice(list(grupo)))
        caracteres_obligatorios.append(caracter_elegido)

    return caracteres_obligatorios
```

### Cuarto Desgranaje

Completar la contraseña

Entrada:

    Crear una función llamada completar_contrasena() que reciba:

    * La longitud que debe tener la contraseña.
    * Los grupos de caracteres seleccionados.
    * Los caracteres obligatorios obtenidos en el tercer desgranaje.

    Su objetivo será añadir caracteres aleatorios hasta alcanzar la longitud solicitada.

Proceso:

    1. Recibir los tres parámetros indicados.

    2. Copiar los caracteres obligatorios en una nueva lista llamada caracteres_contrasena.
    Utilizamos .copy() para crear una lista nueva y conservar sin cambios la lista original de caracteres obligatorios.

    3. Unir los grupos seleccionados para crear un único conjunto de caracteres disponibles con join().

    En el desgranaje anterior necesitábamos mantener los grupos separados para escoger un carácter obligatorio de cada uno.
    Ahora los unimos porque los caracteres restantes pueden pertenecer a cualquiera de los grupos permitidos.

    4. Comprobar si la cantidad de caracteres añadidos todavía es menor que la longitud solicitada.

    5. Seleccionar aleatoriamente un carácter de todos los caracteres disponibles con random.choice().
    NumPy escogerá una mayúscula, una minúscula o un número porque esos son los grupos que hemos incluido.

    6. Añadir el carácter seleccionado a la lista de la contraseña.

    7. Repetir la selección mientras la contraseña no haya alcanzado la longitud solicitada.

    8. Devolver la lista completa de caracteres.

```python
def completar_contrasena(
    longitud: int,
    grupos_seleccionados: list[str],
    caracteres_obligatorios: list[str]
) -> list[str]:

    caracteres_contrasena = caracteres_obligatorios.copy()

    caracteres_disponibles = "".join(grupos_seleccionados)

    while len(caracteres_contrasena) < longitud:
        caracter_elegido = str(
            np.random.choice(list(caracteres_disponibles))
        )
        caracteres_contrasena.append(caracter_elegido)

    return caracteres_contrasena
```

    Su objetivo será cambiar aleatoriamente la posición de todos los caracteres para que los obligatorios no aparezcan siempre al principio.

Proceso:

    1. Crear una copia de la lista "caracteres_contrasena" y utilizamos .copy() para no modificar la lista original. La nueva lista se llamará "caracteres_mezclados".

    2. Mezclar la copia con NumPy. Utlizamos np.random.shuffle() porque cambia aleatoriamente la posición de los elementos de la lista.

```python
def mezclar_contrasena(
    caracteres_contrasena: list[str]
) -> list[str]:

    caracteres_mezclados = caracteres_contrasena.copy()

    np.random.shuffle(caracteres_mezclados)

    return caracteres_mezclados
```

### Séptimo desgranaje

Crear la función principal

Entrada:

    Crear una función llamada crear_contrasena() que conecte y ejecute todas las funciones anteriores en el orden correcto.

    Esta será la función principal u orquestadora. No realizará directamente todo el trabajo, sino que irá solicitando cada tarea a la función correspondiente.

    La función recibirá cinco parámetros:

    longitud, mayusculas, minusculas, numeros, signos

Salida:

    Una contraseña generada como cadena de texto de tipo str. Si la configuración no es válida, devolver None.

Proceso:

1. Llamar a validar_configuracion() para comprobar que los parámetros recibidos permitan generar la contraseña. Si la configuración no es válida, detener el proceso y devolver None.

2. Llamar a obtener_grupos_caracteres() para obtener los grupos correspondientes a los parámetros que tengan el valor True.

3. Llamar a seleccionar_caracteres_obligatorios() para escoger aleatoriamente un carácter de cada grupo seleccionado.

4. Llamar a completar_contrasena() para añadir caracteres aleatorios hasta alcanzar la longitud solicitada.

5. Llamar a mezclar_contrasena() para cambiar aleatoriamente la posición de todos los caracteres.

6. Llamar a formar_contrasena() para unir la lista de caracteres mezclados y convertirla en una única cadena de texto.

7. Devolver la contraseña definitiva.

```python
def crear_contrasena(
    longitud: int,
    mayusculas: bool = True,
    minusculas: bool = True,
    numeros: bool = True,
    signos: bool = False
) -> str | None:

    configuracion_valida = validar_configuracion(
        longitud,
        mayusculas,
        minusculas,
        numeros,
        signos
    )

    if not configuracion_valida:
        return None

    grupos_seleccionados = obtener_grupos_caracteres(
        mayusculas,
        minusculas,
        numeros,
        signos
    )

    caracteres_obligatorios = seleccionar_caracteres_obligatorios(
        grupos_seleccionados
    )

    caracteres_contrasena = completar_contrasena(
        longitud,
        grupos_seleccionados,
        caracteres_obligatorios
    )

    caracteres_mezclados = mezclar_contrasena(
        caracteres_contrasena
    )

    contrasena_final = formar_contrasena(
        caracteres_mezclados
    )

    return contrasena_final
```

    * goles_visitante

Entrada: la función recibirá:

Proceso:

1. Crear una copia del DataFrame recibido para no modificar directamente el original.

```python
def preparar_partidos(partidos: pd.DataFrame) -> pd.DataFrame:

  # Crear una copia para no modificar el DataFrame original
  partidos_preparados = partidos.copy()

  # Separar el resultado por el guion
  goles_separados = (
      partidos_preparados['resultado'].str.split('-', expand=True)
  )

  #Convertir los goles de texto a números enteros
  goles_separados = goles_separados.astype(int)

  # Añadir las dos columnas al DataFrame
  partidos_preparados[
      ['goles_local', 'goles_visitante']] = goles_separados

  return partidos_preparados
```

```python
def crear_registros_equipos(
    partidos_preparados: pd.DataFrame) -> pd.DataFrame:

  # Crear un DataFrame con la perspectiva de los equipos locales
  equipos_locales = pd.DataFrame(
      {'equipo': partidos_preparados['equipo_local'],
       'goles_favor': partidos_preparados['goles_local'],
       'goles_contra': partidos_preparados['goles_visitante'],
       'ganados': (
           partidos_preparados['goles_local'] >
           partidos_preparados['goles_visitante']).astype(int),
       'empatados': (
          partidos_preparados['goles_local'] ==
          partidos_preparados['goles_visitante']).astype(int),
       'perdidos': (
           partidos_preparados['goles_local'] <
           partidos_preparados['goles_visitante']).astype(int)
      })

  # Crear los registros desde la perspectiva del equipo visitante
  equipos_visitantes = pd.DataFrame({
        "equipo": partidos_preparados["equipo_visitante"],
        "goles_favor": partidos_preparados["goles_visitante"],
        "goles_contra": partidos_preparados["goles_local"],
        "ganados": (
            partidos_preparados["goles_visitante"]
            > partidos_preparados["goles_local"]
        ).astype(int),
        "empatados": (
            partidos_preparados["goles_visitante"]
            == partidos_preparados["goles_local"]
        ).astype(int),
        "perdidos": (
            partidos_preparados["goles_visitante"]
            < partidos_preparados["goles_local"]
        ).astype(int)
    })

  # Unir los registros locales y visitantes
  registros_equipos = pd.concat(
      [equipos_locales, equipos_visitantes],
      ignore_index=True
  )

  return registros_equipos
```

```python
# Comprobación de la función crear_registros_equipos

registros_equipos_prueba = crear_registros_equipos(
    partidos_preparados_prueba
)

# Mostrar las primeras cinco filas
display(registros_equipos_prueba.head())

# Comprobar la cantidad total de registros
print(
    "Cantidad de registros:",
    len(registros_equipos_prueba)
)

# Comprobar los nombres de las columnas
print(
    "Nombres de las columnas:",
    registros_equipos_prueba.columns.tolist()
)

# Obtener la posición del primer equipo visitante
# Recordar que partidos_preparados_prueba tiene 500 filas,
# en cambio registros_equipos_prueba tiene 1000.
posicion_primer_visitante = len(
    partidos_preparados_prueba
)

# Mostrar las dos perspectivas del primer partido
print("\nPerspectivas del primer partido:")

display(
    registros_equipos_prueba.iloc[
        [0, posicion_primer_visitante]
    ]
)
```

### Cuarto desgranaje:

Calcular las estadísticas de cada equipo.

Objetivo

    Crear una función llamada "calcular_estadisticas()" que agrupe todos los registros correspondientes al mismo equipo y sume sus estadísticas.

    Actualmente "registros_equipos_prueba" contiene 1000 filas. Un mismo equipo aparece muchas veces porque tiene un registro por cada partido jugado.

Entrada:La función recibirá:

    "registros_equipos": el DataFrame creado en el tercer desgranaje.

Proceso:

    1. Agrupar las filas por la columna equipo.

    2. Seleccionar las columnas numéricas que queremos sumar.

    3. Sumar las estadísticas de todos los partidos de cada equipo.

    4. Crear la columna puntos.

    5. Calcular los puntos mediante:
        ganados × 3 + empatados
    
    6. Devolver el DataFrame de estadísticas.

Salida:

La función devolverá un DataFrame con una fila por equipo y estas columnas:

    goles_favor
    goles_contra
    ganados
    empatados
    perdidos
    puntos

```python
def calcular_estadisticas(registros_equipos: pd.DataFrame) -> pd.DataFrame:

  # Agrupar los registros por equipo y sumar las estadísticas
  estadisticas_equipos = (
      registros_equipos.groupby('equipo')[
          [
              'goles_favor',
              'goles_contra',
              'ganados',
              'empatados',
              'perdidos'
          ]
      ].sum()
  )

  # Calcular los puntos de cada equipo
  estadisticas_equipos["puntos"] = (
      estadisticas_equipos["ganados"] * 3
     + estadisticas_equipos["empatados"]
    )

  return estadisticas_equipos
```

```python
# Comprobación de la función calcular_estadisticas

estadisticas_equipos_prueba = calcular_estadisticas(
    registros_equipos_prueba
)

# Mostrar las primeras cinco filas
display(estadisticas_equipos_prueba)

# Comprobar la cantidad de equipos
print(
    "Cantidad de equipos:",
    len(estadisticas_equipos_prueba)
)

# Comprobar los nombres de las columnas
print(
    "Nombres de las columnas:",
    estadisticas_equipos_prueba.columns.tolist()
)

# Comprobar el nombre del índice
print(
    "Nombre del índice:",
    estadisticas_equipos_prueba.index.name
)
```

```python
def ordenar_clasificacion(
    estadisticas_equipos:pd.DataFrame) -> pd.DataFrame:

    clasificacion = estadisticas_equipos.sort_values(
        by='puntos', ascending=False
    )
    return clasificacion
```

### Noveno desgranaje:

***Función principal***

Objetivo:

    Crear una función llamada "main_futbol()" que conecte todas las funciones anteriores en el orden correcto.

    Esta función actuará como orquestadora.

    Cargar
      ↓
    Preparar
      ↓
    Crear registros
      ↓
    Calcular estadísticas
      ↓
    Obtener equipos destacados
      ↓
    Ordenar clasificación
      ↓
    Mostrar resultados

Entrada:

    Recibirá ruta_archivo: str
    Será la ubicación del archivo historic_partits.txt.

Proceso:

1. Cargar el archivo.
2. Preparar los resultados.
3. Crear los registros de locales y visitantes.
4. Calcular las estadísticas acumuladas.
5. Obtener el equipo más goleador.
6. Obtener el equipo más goleado.
7. Ordenar la clasificación.
8. Mostrar los resultados.
9. Devolver la clasificación final.

Salida:

    Devolverá la clasificación final como un DataFrame

```python
def main_futbol(
    ruta_archivo: str
) -> pd.DataFrame:

    # Cargar los partidos desde el archivo
    partidos = cargar_partidos(
        ruta_archivo
    )

    # Separar los resultados en goles locales y visitantes
    partidos_preparados = preparar_partidos(
        partidos
    )

    # Crear una fila para cada equipo participante en cada partido
    registros_equipos = crear_registros_equipos(
        partidos_preparados
    )

    # Calcular las estadísticas acumuladas (incluido el cálculo de puntos)
    estadisticas_equipos = calcular_estadisticas(
        registros_equipos
    )

    # Obtener el equipo más goleador
    equipo_mas_goleador = obtener_equipo_mas_goleador(
        estadisticas_equipos
    )

    # Obtener el equipo más goleado
    equipo_mas_goleado = obtener_equipo_mas_goleado(
        estadisticas_equipos
    )

    # Ordenar la clasificación por puntos de mayor a menor
    clasificacion = ordenar_clasificacion(
        estadisticas_equipos
    )

    # Mostrar los resultados finales
    mostrar_resultados(
        equipo_mas_goleador,
        equipo_mas_goleado,
        clasificacion
    )

    # Devolver la clasificación
    return clasificacion
```
